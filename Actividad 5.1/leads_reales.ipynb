```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Cargar archivo csv desde seaborn
df=pd.read_csv("Leads_Reales_sin_valores_atipicos.csv")
df.head(5)
```

```python
# Eliminar la columna 'Unnamed: 0' del DataFrame
data = df.drop(columns=['Unnamed: 0'])
data.head(5)
```

```python
# Mes 
data['Mes'] = data['Mes'].str.strip().str.lower()
data['Mes'] = data['Mes'].str.replace('enero', '1').replace('febrero', '2').replace('marzo', '3').replace('abril', '4').replace('mayo', '5').replace('junio', '6').replace('julio', '7').replace('agosto', '8').replace('septiembre', '9').replace('octubre', '10').replace('noviembre', '11').replace('diciembre', '12')
```

```python
# Año
data['Año'] = data['Año'].astype(int)
data
```

```python
# Graficamos todas las dispersiones entre todas las variables
sns.pairplot(data)
```

```python
#Encontramos todas las correlaciones entre las variables 
Corr_Factors=data.corr()
Corr_Factors
```

```python
#Encontramos el valor absoluto de todas las correlaciones entre las variables 
Corr_Factors1= abs(Corr_Factors)
Corr_Factors1
```

```python
#Graficamos el mapa de calor de los coeficientes de correlación 
Heat_Map= sns.heatmap(Corr_Factors1, cmap = 'coolwarm')  
Heat_Map
```

```python
#Ajustamos el mapa de calor de los coeficientes de correlación 
Heat_Map= sns.heatmap(Corr_Factors1, cmap = 'Greens', annot=True, fmt=".2f")  
Heat_Map
```

```python
# Imprimimos el scatter plot entre la variable dependiente (Ventas) e independiente (Efectivos) para observar el comportamiento en su dispersión 
from turtle import color
sns.scatterplot(x='Efectivos', y='Total', color="blue", data=data)
```

```python
# Declaramos las variables dependientes e independientes para la regresión lineal
Vars_Indep= data[['Efectivos']]
Var_Dep= data['Total']
```

```python
%pip install scikit-learn
```

```python
# Se define model como la función de regresión lineal
from sklearn.linear_model import LinearRegression
model= LinearRegression()
```

```python
# Ajustamos el modelo con las variables antes declaradas
model.fit(X=Vars_Indep, y=Var_Dep)
```

```python
# Verificamos los coeficientes obtenidos para el modelo ajustado
model.__dict__
```

```python
# Evaluamos la eficiencia del modelo obtenido por medio del coeficiente R Determinación
model.score(Vars_Indep,Var_Dep)
```

```python
# Predecimos los valores de ventas a partir de la variable "efectivos" y "conversión"
y_pred= model.predict(X=data[['Efectivos']])
y_pred
```

```python
# Insertamos la columna de predicciones en el DataFrame
data.insert(0, 'Predicciones', y_pred)
data
```

```python
#Visualizamos la gráfica comparativa entre el total real y el total predecido
sns.scatterplot(x='Efectivos', y='Total', color="blue", data=data)
sns.scatterplot(x='Efectivos', y='Predicciones', color="red", data=data)
sns.lineplot(x='Efectivos', y='Predicciones', color="red", data=data)
```

```python
# Corroboramos cual es el coeficiente de Determinación de nuestro modelo
coef_Deter=model.score(X=Vars_Indep, y=Var_Dep)
coef_Deter
```

```python
# Corroboramos cual es el coeficiente de Correlación de nuestro modelo
coef_Correl=np.sqrt(coef_Deter)
coef_Correl
```
